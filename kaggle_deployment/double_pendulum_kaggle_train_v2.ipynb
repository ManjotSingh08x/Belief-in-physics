# Double Pendulum Production Training & Multi-Platform Queue Architecture (v2)

This notebook implements the v2 production training and GPU-accelerated belief probing pipeline for the chaotic, Lagrangian **Double Pendulum** (`double_pendulum_mess4`).

---

## v2 Architecture Specifications

1. **Phase 1: Physics Screening & 5-Tuple Distillation**:
   - **Sweep 1A**: Grid screening $(\Delta v \times n)$ under baseline contractive damping $\gamma = 0.65$ ($dt = 0.2$, $m = 5000 // n$).
   - **Intermediate Step**: Steady-state energy balance $\gamma^*$ calculation via `compute_optimal_gamma`, generating candidate gammas ($k \in \{0.03, 0.07, 0.10\}, d\gamma = 0.05$).
   - **Sweep 1B**: Damping sweep on candidate pairs.
   - **Distillation Checkpoint 2**: Diversity-constrained selection of the **top 5 physics tuples** (`distilled_physics_tuples.csv` & `.json`).
2. **Phase 2: 15-Model Production Queue (Single Model per Configuration, Zero Seeds)**:
   - Factorial matrix: **5 physics tuples** $\times$ **1 fixed width** ($d_{\text{model}} = 128$) $\times$ **3 lookaheads** (`_k1`, `_k_mid`, `_k_full`) $\times$ **1 model/config** = **15 models total**.
   - **Randomized Trajectory Starting Velocity**: Every trajectory draws $(\omega_1, \omega_2) \sim \text{truncnorm}([-\Delta v, \Delta v], \mu=0, \sigma^2 = \Delta v / 2)$.
   - **GPU-Accelerated Probing**: Layer-wise Ridge regression across 2 techniques (all_tokens_probing & usable_tokens_probing) solved via PyTorch closed-form linear algebra directly on device (100x faster than CPU Scikit-learn).
   - **Locked Batch Size Invariant**: `BATCH_SIZE = 1024` with micro-batching and gradient accumulation.
   - **Deep Token Budget**: 800M tokens/model ($156$ gradient steps); smoke mode: 10,000 tokens.
   - **Clean Naming**: Model IDs `{exp_name}_{train_name}` without seed suffixes.
3. **Phase 3: Emergence Synthesis & 5-Panel Dashboard**:
   - Consolidated `master_probe_results.csv` (omitting seed column).
   - Adapted 5-Panel Dashboard:
     1. `scaling_curves.png`: Config performance comparison across the 5 physics configurations ($d_{\text{model}}=128$)
     2. `lookahead_comparison.png`: Forecasting horizon comparison ($k=1$ vs $k=\lfloor n/2 \rfloor$ vs $k=n$)
     3. `probe_views_breakdown.png`: Layer-wise linear readability breakdown (all_tokens_probing vs usable_tokens_probing)
     4. `physics_vs_belief_scatter.png`: Representation decoupling scatter (15 models)
     5. `trained_vs_random_gain.png`: Net emergence gain isolating non-trivial feature learning
   - Master distribution package: `double_pendulum_results.zip`.

## Phase 2: The 15-Model Production Queue (Single Model per Configuration, Zero Seeds)

The production queue trains:
$$\text{Total Models} = 5\text{ physics tuples} \times 1\text{ width } (128) \times 3\text{ lookaheads } (k \in \{1, \lfloor n/2 \rfloor, n\}) \times 1\text{ model} = \mathbf{15\text{ models}}$$

- **Randomized Initial Velocities**: Drawn from $\text{truncnorm}([-\Delta v, \Delta v], \mu=0, \sigma^2 = \Delta v / 2)$ for every trajectory.
- **GPU-Accelerated In-Worker Probing**: Instant closed-form Ridge regression solved directly on GPU.
- **Batch Size Invariant**: `BATCH_SIZE = 1024` with micro-batching (`micro_batch = 64`).
- **Token Budget**: 800M tokens/model; smoke mode: 10,000 tokens.


```python
# ==============================================================================
# Cell 3: Phase 2 - Production Model Queue Execution (15 Models, GPU Probing)
# ==============================================================================
import concurrent.futures
from scripts.production_train_v2 import (
    LookaheadTransformer,
    train_and_probe_worker,
    sync_from_kaggle_input,
)

tokens_per_model = 10_000 if SMOKE else 800_000_000
batch_size = 1024  # strictly locked invariant

w = 128
d_mlp = 4 * w

jobs: List[ModelJob] = []
for exp_cfg in chosen_experiment_configs:
    n = exp_cfg.n
    lookahead_specs = [
        (1, "_k1"),
        (max(1, n // 2), "_k_mid"),
        (n, "_k_full"),
    ]
    for k_val, k_suf in lookahead_specs:
        train_name = f"dmodel_{w}_dmlp_{d_mlp}{k_suf}"
        train_cfg = TrainingConfig(
            name=train_name,
            d_model=w,
            d_mlp=d_mlp,
            n_layers=4,
            n_heads=2,
            k=k_val,
            k_suffix=k_suf,
        )
        j_id = f"{exp_cfg.name}_{train_name}"
        job = ModelJob(
            job_id=j_id,
            exp_config=exp_cfg,
            train_config=train_cfg,
            k=k_val,
            k_suffix=k_suf,
            tokens_seen=0,
            target_tokens=tokens_per_model,
            checkpoint_path=RESULTS_DIR / j_id / f"{j_id}_final.pt",
            sidecar_path=RESULTS_DIR / j_id / f"{j_id}_final.json",
            probe_csv_path=RESULTS_DIR / j_id / f"{j_id}_probes.csv",
            prob_map_path=RESULTS_DIR / j_id / f"{j_id}_lookahead_prob_map.png",
            status="QUEUED",
            seed=20260925,
        )
        jobs.append(job)

if SMOKE:
    jobs = jobs[:2]

print(f"\n==============================================================================")
print(f" PRODUCTION MODEL QUEUE ({len(jobs)} TOTAL MODELS)")
print(f"==============================================================================")
print(f"Queue Depth: {len(jobs)} models across {len(chosen_experiment_configs)} physics tuples at fixed width {w}.")

manifest_path = OUTPUT_BASE / "pipeline_state.json"
master_csv_path = OUTPUT_BASE / "master_probe_results.csv"

manifest_data = {
    "platform_mode": PLATFORM_MODE,
    "completed_models": [],
    "job_statuses": {},
    "total_models": len(jobs),
    "physics_tuples": [cfg.name for cfg in chosen_experiment_configs],
    "total_tokens_per_model": tokens_per_model,
    "timestamp": time.time(),
}

all_probe_dfs: List[pd.DataFrame] = []
if PLATFORM_MODE == "kaggle":
    sync_from_kaggle_input(RESULTS_DIR)

if master_csv_path.exists():
    try:
        existing_df = pd.read_csv(master_csv_path)
        all_probe_dfs.append(existing_df)
        completed_set = set(existing_df["model_id"].unique())
        print(f"Auto-Resume: Loaded {len(completed_set)} previously completed models from {master_csv_path.name}.")
    except Exception as e:
        print(f"Notice: Could not parse existing master CSV ({e}); starting fresh.")
        completed_set = set()
else:
    completed_set = set()

for job in jobs:
    zip_path = RESULTS_DIR / f"{job.job_id}.zip"
    if job.job_id in completed_set or (zip_path.exists() and job.probe_csv_path.exists()):
        job.status = "COMPLETED"
        if job.job_id not in manifest_data["completed_models"]:
            manifest_data["completed_models"].append(job.job_id)
        if job.probe_csv_path.exists() and (not completed_set or job.job_id not in completed_set):
            try:
                all_probe_dfs.append(pd.read_csv(job.probe_csv_path))
            except Exception:
                pass
    manifest_data["job_statuses"][job.job_id] = job.status

with open(manifest_path, "w") as f:
    json.dump(manifest_data, f, indent=2)

# Execution Dispatch (GPU Probing)
if NUM_TRAIN_WORKERS > 1:
    print(f"\nDispatching {len(jobs)} jobs across {NUM_TRAIN_WORKERS} concurrent GPU workers ({WORKER_DEVICES})...")
    ctx = torch.multiprocessing.get_context("spawn")
    with concurrent.futures.ProcessPoolExecutor(max_workers=NUM_TRAIN_WORKERS, mp_context=ctx) as executor:
        active_futs = {}
        for i, job in enumerate(jobs):
            if job.status == "COMPLETED":
                continue
            dev = WORKER_DEVICES[i % len(WORKER_DEVICES)]
            job.status = "TRAINING"
            manifest_data["job_statuses"][job.job_id] = job.status
            fut = executor.submit(train_and_probe_worker, job=job, total_tokens=tokens_per_model, batch_size=batch_size, device=dev, results_dir=RESULTS_DIR, smoke=SMOKE)
            active_futs[fut] = job

        for fut in concurrent.futures.as_completed(active_futs):
            job = active_futs[fut]
            job_id, probe_df = fut.result()
            job.status = "COMPLETED"
            if not probe_df.empty:
                all_probe_dfs.append(probe_df)
                cur_master_df = pd.concat(all_probe_dfs, ignore_index=True)
                cur_master_df.to_csv(master_csv_path, index=False)
            if job.job_id not in manifest_data["completed_models"]:
                manifest_data["completed_models"].append(job.job_id)
            manifest_data["job_statuses"][job.job_id] = job.status
            manifest_data["timestamp"] = time.time()
            with open(manifest_path, "w") as f:
                json.dump(manifest_data, f, indent=2)
            print(f"-> Progress saved: {len(manifest_data['completed_models'])}/{len(jobs)} completed.", flush=True)
else:
    dev = WORKER_DEVICES[0]
    for job in jobs:
        if job.status == "COMPLETED":
            continue
        job.status = "TRAINING"
        manifest_data["job_statuses"][job.job_id] = job.status
        _, probe_df = train_and_probe_worker(job=job, total_tokens=tokens_per_model, batch_size=batch_size, device=dev, results_dir=RESULTS_DIR, smoke=SMOKE)
        job.status = "COMPLETED"
        if not probe_df.empty:
            all_probe_dfs.append(probe_df)
            cur_master_df = pd.concat(all_probe_dfs, ignore_index=True)
            cur_master_df.to_csv(master_csv_path, index=False)
        if job.job_id not in manifest_data["completed_models"]:
            manifest_data["completed_models"].append(job.job_id)
        manifest_data["job_statuses"][job.job_id] = job.status
        manifest_data["timestamp"] = time.time()
        with open(manifest_path, "w") as f:
            json.dump(manifest_data, f, indent=2)
        print(f"-> Progress saved: {len(manifest_data['completed_models'])}/{len(jobs)} completed.", flush=True)
```
